```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
spark = SparkSession.builder.appName("MultipleDatasetApp") \
        .master("local[4]") \
        .config("spark.dynamicAllocation.enabled", "false") \
        .config("spark.sql.adaptive.enabled", "false") \
        .getOrCreate()
```

```python
sc = spark.sparkContext
```

```python
sc
```

```python
driversDF = spark.read \
                .option("header", "true") \
                .option("inferschema", "true") \
                .csv("/home/luffy/Documents/Spark/Data/Drivers.csv")
```

```python
driversDF.createOrReplaceTempView("Drivers")
```

```python
driversDF.printSchema()
```

```python
driversDF.show()
```

```python
cabsDF = spark.read \
                .option("header", "true") \
                .option("inferschema", "true") \
                .csv("/home/luffy/Documents/Spark/Data/Cabs.csv")
```

```python
cabsDF.printSchema()
```

```python
cabsDF.createOrReplaceTempView("Cabs")
```

```python
spark.sql("""

SELECT Name FROM Cabs WHERE LicenseType = "OWNER MUST DRIVE"
UNION
SELECT Name FROM Drivers
""").count()
```

```python
spark.sql("""

SELECT Name FROM Cabs WHERE LicenseType = "OWNER MUST DRIVE"
INTERSECT
SELECT Name FROM Drivers
""").count()
```
