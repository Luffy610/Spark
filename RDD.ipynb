```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName("RDD").master("local[*]").getOrCreate()
```

```python
spark
```

```python
sc = spark.sparkContext
```

```python
sc
```

```python
numbersRdd = sc.parallelize([1,2,3,4,5])
```

```python
numbersRdd.getNumPartitions()
```

```python
output = numbersRdd.collect()
print(output)
```

```python
numbersRdd.take(2)
```

```python
carsData = sc.textFile("/home/luffy/Documents/Spark/Data/cars24data.csv", 4)
```

```python
carsData.take(10)
```

```python
carsData.getNumPartitions()
```

```python
header = carsData.first()
carsDatawithCols = carsData.filter(lambda row: row != header)
```

```python
carsDatawithCols = carsDatawithCols.map (lambda row: row.split(","))
```

```python
carsDatawithCols.take(3)
```

```python
carsDatawithCols.filter(lambda row: "Manual" in row).take(5)
```

## Pair RDD

```python
import math
```

```python
numbersRDD = sc.parallelize([2,3,4,5,6])
```

```python
numsWithSquareRootsRdd = numbersRDD.map(lambda num: (num, math.sqrt(num)))
```

```python
numsWithSquareRootsRdd.collect()
```

```python
carsPairRDD = carsDatawithCols.map(lambda row: (row[2], row[0]))
```

```python
carsPairRDD.take(5)
```
